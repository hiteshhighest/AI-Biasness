## Gender Distribution

The distribution of candidates across gender categories is examined to assess the composition of the dataset.

```python
df["Gender"].value_counts(dropna=False)
```

The distribution of candidates across gender categories is visualised using a bar chart.

```python
df["Gender"].value_counts().plot(kind="bar")

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Candidates")
plt.xticks(rotation=0)

plt.show()
```

## Education Level Distribution

 The distribution of candidates across education levels is examined to understand the composition of the dataset.

```python
df["Education_Level"].value_counts(dropna=False)
```

The distribution of candidates across education levels is visualised using a bar chart.

```python
df["Education_Level"].value_counts().plot(kind="bar")

plt.title("Education Level Distribution")
plt.xlabel("Education Level")
plt.ylabel("Number of Candidates")
plt.xticks(rotation=45)

plt.show()
```

## Age Distribution

The age distribution of candidates is examined to understand the range and central tendency of the dataset.

```python
df["Age"].describe()
```

The distribution of candidate ages is visualised using a histogram.

```python
plt.hist(df["Age"], bins=20)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Candidates")

plt.show()
```

## Experience Distribution

## Candidate Score Distributions

The descriptive statistics of the candidate assessment scores are examined to understand their ranges and distributions.

```python
score_columns = [
    "Skill_Score",
    "Aptitude_Test_Score",
    "Technical_Test_Score",
    "Communication_Score",
    "Interview_Score"
]

df[score_columns].describe()
```

The distributions of the five candidate assessment scores are visualised to compare their ranges and overall patterns.

```python
for column in score_columns:
    plt.hist(df[column], bins=20)
    
    plt.title(f"{column} Distribution")
    plt.xlabel("Score")
    plt.ylabel("Number of Candidates")
    
    plt.show()
```
